```python
def add_one(x):
    return x+1
```

```python
from langchain_core.runnables import RunnableLambda
r = RunnableLambda(add_one)
r.invoke(3)
```

```python
r = RunnableLambda(lambda x:x+1)
r.invoke(4)
```

```python
#chain of runnables called sequence
sequence = RunnableLambda(lambda x: x+2) | RunnableLambda(lambda x: x*2)
sequence.invoke(2)
```

```python
#for multiple values
sequence.batch([1,2,3])
```

```python
from langchain_core.globals import set_debug
set_debug(True  )
```

```python
from langchain_core.runnables import RunnableParallel

rp = RunnableParallel(
    {
        "addone":RunnableLambda(add_one),
        "multiply":RunnableLambda(lambda x: x*2)
    }
)
rp.invoke(2)
```
